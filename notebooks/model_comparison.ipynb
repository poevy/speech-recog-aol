# Model Comparison: Code-Switching ASR

This notebook compares fine-tuned ASR models for Indonesian-English code-switching.

## Models Compared
1. **Whisper-base** (baseline, from original notebook)
2. **Whisper-tiny** (speed-focused)
3. **Wav2Vec2-Large-XLSR** (Indonesian-specialized, different architecture)

## Metrics
- WER (Word Error Rate)
- CER (Character Error Rate)
- RTF (Real-Time Factor)
- Latency
- Code-switching accuracy

```python
# Setup
import sys
sys.path.insert(0, '..')

import os
import json
import numpy as np
import matplotlib.pyplot as plt
import torch

from src.data_preparation import get_datasets, get_sample_batch
from src.metrics import evaluate_model, calculate_wer, calculate_cer
from src.visualization import *

print(f"PyTorch: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
```

## 1. Load Datasets

```python
# Load datasets
datasets = get_datasets(
    cs_train_size=1000,
    cs_test_size=150,
    verbose=True
)

test_sets = datasets['test_sets']
print(f"\nTest sets: {list(test_sets.keys())}")
```

## 2. Load Models

```python
# Model paths - update these after training
MODEL_PATHS = {
    'whisper-base': '../outputs/whisper_base/final',  # From original notebook
    'whisper-tiny': '../outputs/whisper_tiny/final',
    'wav2vec2-large': '../outputs/wav2vec2/final'
}

# Check which models are available
available_models = {}
for name, path in MODEL_PATHS.items():
    if os.path.exists(path):
        available_models[name] = path
        print(f"✓ {name}: Found")
    else:
        print(f"✗ {name}: Not found at {path}")

print(f"\n{len(available_models)} models available for comparison")
```

## 3. Evaluate Models

```python
# Load and evaluate each model
results = {}

for model_name, model_path in available_models.items():
    print(f"\nEvaluating {model_name}...")
    
    # Load model based on type
    if 'whisper' in model_name:
        from transformers import WhisperForConditionalGeneration, WhisperProcessor
        model = WhisperForConditionalGeneration.from_pretrained(model_path)
        processor = WhisperProcessor.from_pretrained(model_path)
        device = 'cuda' if torch.cuda.is_available() else 'cpu'
        model = model.to(device)
        
        def model_fn(audio):
            inputs = processor(audio, sampling_rate=16000, return_tensors="pt")
            inputs = {k: v.to(device) for k, v in inputs.items()}
            pred_ids = model.generate(**inputs, max_new_tokens=128)
            return processor.decode(pred_ids[0], skip_special_tokens=True)
    
    else:  # wav2vec2
        from transformers import Wav2Vec2ForCTC, Wav2Vec2Processor
        model = Wav2Vec2ForCTC.from_pretrained(model_path)
        processor = Wav2Vec2Processor.from_pretrained(model_path)
        device = 'cuda' if torch.cuda.is_available() else 'cpu'
        model = model.to(device)
        
        def model_fn(audio):
            inputs = processor(audio, sampling_rate=16000, return_tensors="pt")
            inputs = {k: v.to(device) for k, v in inputs.items()}
            with torch.no_grad():
                logits = model(**inputs).logits
            pred_ids = torch.argmax(logits, dim=-1)
            return processor.decode(pred_ids[0])
    
    # Evaluate on test sets
    model_results = {}
    for test_name, test_ds in test_sets.items():
        print(f"  Evaluating on {test_name}...")
        eval_results = evaluate_model(model_fn, test_ds, max_samples=min(50, len(test_ds)))
        model_results[test_name] = eval_results
    
    results[model_name] = model_results

print("\nEvaluation complete!")
```

## 4. Results Visualizations

```python
# Extract metrics for comparison
wer_comparison = {}
speed_comparison = {}

for model_name, model_results in results.items():
    wer_comparison[model_name] = {
        test_set: res['wer'] for test_set, res in model_results.items()
    }
    speed_comparison[model_name] = {
        'rtf': np.mean([res['rtf'] for res in model_results.values()]),
        'avg_latency': np.mean([res['avg_latency'] for res in model_results.values()])
    }

print("WER Comparison:")
print(json.dumps(wer_comparison, indent=2))
```

```python
# Plot WER comparison
fig = plot_wer_comparison(wer_comparison)
plt.show()
```

```python
# Plot accuracy vs speed tradeoff
overall_results = {}
for model_name, model_results in results.items():
    avg_wer = np.mean([res['wer'] for res in model_results.values()])
    avg_rtf = np.mean([res['rtf'] for res in model_results.values()])
    overall_results[model_name] = {'wer': avg_wer, 'rtf': avg_rtf}

fig = plot_accuracy_speed_tradeoff(overall_results)
plt.show()
```
