# Code-Switching ASR Training - Google Colab

This notebook trains Whisper-tiny and Wav2Vec2 models for Indonesian-English code-switching ASR.

## Training Settings (matching original Whisper-base)
- MAX_STEPS: 1500 (same as original)
- Batch size: 16
- Learning rate: 1e-5
- Dataset: FLEURS ID + EN + synthetic code-switching (same as original)

## Instructions
1. Set runtime to GPU: Runtime > Change runtime type > T4 GPU
2. Run all cells

```python
# ======================= CONFIGURATION =======================
# Settings match original Whisper-base training
GITHUB_REPO = "https://github.com/poevy/speech-recog-aol.git"

# Training parameters (SAME AS ORIGINAL)
MAX_STEPS = 1500          # Same as original MAX_STEPS
BATCH_SIZE = 16           # Same as original
LEARNING_RATE = 1e-5      # Same as original
WARMUP_STEPS = 100        # Same as original
EVAL_STEPS = 500          # Same as original

# Dataset sizes (SAME AS ORIGINAL)
CS_TRAIN_SIZE = 1000      # Same as original
CS_TEST_SIZE = 150        # Same as original

# Set to True to train each model
TRAIN_WHISPER_TINY = True
TRAIN_WAV2VEC2 = True
# =============================================================
```

## 3. Train Whisper-Tiny (Settings match original Whisper-base)

```python
if TRAIN_WHISPER_TINY:
    print("="*60)
    print("Training Whisper-Tiny...")
    print(f"MAX_STEPS: {MAX_STEPS}, BATCH_SIZE: {BATCH_SIZE}, LR: {LEARNING_RATE}")
    print("="*60)
    
    !python models/train_whisper_tiny.py \
        --output_dir outputs/whisper_tiny \
        --max_steps {MAX_STEPS} \
        --cs_train_size {CS_TRAIN_SIZE} \
        --cs_test_size {CS_TEST_SIZE} \
        --batch_size {BATCH_SIZE} \
        --learning_rate {LEARNING_RATE} \
        --eval_steps {EVAL_STEPS}
    
    print("\nWhisper-tiny training complete!")
else:
    print("Skipping Whisper-tiny training")
```

## 4. Train Wav2Vec2

```python
if TRAIN_WAV2VEC2:
    print("="*60)
    print("Training Wav2Vec2...")
    print(f"MAX_STEPS: {MAX_STEPS}, BATCH_SIZE: 8, LR: 3e-5")
    print("="*60)
    
    !python models/train_wav2vec2.py \
        --output_dir outputs/wav2vec2 \
        --max_steps {MAX_STEPS} \
        --cs_train_size {CS_TRAIN_SIZE} \
        --cs_test_size {CS_TEST_SIZE} \
        --batch_size 8 \
        --learning_rate 3e-5 \
        --eval_steps {EVAL_STEPS}
    
    print("\nWav2Vec2 training complete!")
else:
    print("Skipping Wav2Vec2 training")
```

## 5. View Training Results

```python
import json
import os

print("="*60)
print("TRAINING RESULTS")
print("="*60)

# Whisper-tiny results
whisper_results_path = "outputs/whisper_tiny/final_results.json"
if os.path.exists(whisper_results_path):
    with open(whisper_results_path) as f:
        results = json.load(f)
    print(f"\nWhisper-tiny:")
    print(f"  Training time: {results.get('training_time_minutes', 0):.1f} minutes")
    if 'eval_results' in results:
        for k, v in results['eval_results'].items():
            print(f"  {k}: {v}")

# Wav2Vec2 results
wav2vec2_results_path = "outputs/wav2vec2/final_results.json"
if os.path.exists(wav2vec2_results_path):
    with open(wav2vec2_results_path) as f:
        results = json.load(f)
    print(f"\nWav2Vec2:")
    print(f"  Training time: {results.get('training_time_minutes', 0):.1f} minutes")
    if 'eval_results' in results:
        for k, v in results['eval_results'].items():
            print(f"  {k}: {v}")
```
